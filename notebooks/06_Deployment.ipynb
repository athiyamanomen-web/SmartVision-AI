````python
# ============================================================
# PHASE 6 - STEP 6.1
# CELL 2 - SCAN STREAMLIT APP DEPENDENCIES
# ============================================================

from pathlib import Path
import ast

PROJECT_DIR = Path(r"E:\SmartVision-AI")
APP_DIR = PROJECT_DIR / "streamlit_app"

python_files = [
    APP_DIR / "app.py",
    *sorted((APP_DIR / "pages").glob("*.py")),
    *sorted((APP_DIR / "utils").glob("*.py")),
]

all_imports = set()

print("=" * 70)
print("SMARTVISION AI - DEPENDENCY SCAN")
print("=" * 70)

for file in python_files:

    print(f"\n📄 {file.relative_to(PROJECT_DIR)}")

    try:
        source = file.read_text(encoding="utf-8")
        tree = ast.parse(source)

        file_imports = set()

        for node in ast.walk(tree):

            if isinstance(node, ast.Import):
                for alias in node.names:
                    file_imports.add(
                        alias.name.split(".")[0]
                    )

            elif isinstance(node, ast.ImportFrom):

                if node.module:
                    file_imports.add(
                        node.module.split(".")[0]
                    )

        all_imports.update(file_imports)

        for package in sorted(file_imports):
            print(f"   • {package}")

    except Exception as e:
        print(f"   ❌ Could not scan: {e}")


print("\n" + "=" * 70)
print("UNIQUE IMPORTS FOUND")
print("=" * 70)

for package in sorted(all_imports):
    print(f"• {package}")

print("\nTotal unique imports:", len(all_imports))

print("\n" + "=" * 70)
print("SCAN COMPLETE")
print("=" * 70)
````

````python
# ============================================================
# PHASE 6 - STEP 6.1
# CELL 11 - CHECK GIT AND GIT LFS
# ============================================================

import subprocess

print("=" * 70)
print("SMARTVISION AI - GIT / GIT LFS CHECK")
print("=" * 70)


def check_command(command, name):

    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            shell=True
        )

        output = (
            result.stdout.strip()
            or result.stderr.strip()
        )

        if result.returncode == 0:
            print(f"\n✅ {name} is installed")
            print(f"   {output}")
            return True

        else:
            print(f"\n❌ {name} is not available")
            print(f"   {output}")
            return False

    except Exception as e:
        print(f"\n❌ Could not check {name}")
        print(f"   {e}")
        return False


git_ok = check_command(
    "git --version",
    "Git"
)

lfs_ok = check_command(
    "git lfs version",
    "Git LFS"
)


print("\n" + "=" * 70)
print("RESULT")
print("=" * 70)

print(f"Git     : {'✅ READY' if git_ok else '❌ NOT READY'}")
print(f"Git LFS : {'✅ READY' if lfs_ok else '❌ NOT READY'}")

if git_ok and lfs_ok:
    print(
        "\n✅ System is ready for Git + Git LFS repository setup."
    )
else:
    print(
        "\n⏳ Missing component must be installed before repository setup."
    )

print("=" * 70)
````
